## 1. Variables and data types

A **variable** is a name that stores a value. Python figures out the **type** of value automatically.

| Type | Example | Used for |
|---|---|---|
| `int` | `5` | whole numbers |
| `float` | `12.50` | decimal numbers |
| `str` | `"Python Basics"` | text |
| `bool` | `True` | yes or no |

````python
title = "Python Basics"     # str
price = 12.50               # float
quantity = 5                # int
in_stock = True             # bool

print(title, price, quantity, in_stock)
print(type(title), type(price), type(quantity), type(in_stock))
````

## 2. Expressions and operators

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+ - * /` | add, subtract, multiply, divide | `7 / 2` | `3.5` |
| `//` | floor division (whole part) | `7 // 2` | `3` |
| `%` | remainder | `7 % 2` | `1` |
| `**` | power | `2 ** 3` | `8` |

`round(value, 2)` rounds to two decimal places, which is useful for money.

````python
subtotal = price * quantity
tax = subtotal * 0.0825
print("Subtotal:", subtotal)
print("Tax:", round(tax, 2))

# Preview of f-strings (you will learn these properly in Week 2):
print(f"Total with tax: ${subtotal + tax:.2f}")
````

## 3. Input and output

`input()` always returns **text** (a string). To do math with it, convert it with `int()` or `float()`.

````python
copies = input("How many copies? ")       # this is a string, e.g. "3"
copies = int(copies)                        # now it is a number
print("You ordered", copies, "copies.")
print("Next order will be", copies + 1, "copies.")
````

## 4. Making decisions: if, elif, else

Python runs the **first** branch whose condition is `True`, then skips the rest. Indentation (4 spaces) shows which lines belong to each branch.

Comparison operators: `==  !=  <  <=  >  >=`. Combine conditions with `and`, `or`, `not`.

````python
stock = 3

if stock == 0:
    print("Out of stock")
elif stock < 5:
    print("Low stock: reorder soon")
else:
    print("In stock")
````

## 5. Repeating actions: loops

A **`for` loop** repeats a set number of times. A **`while` loop** repeats as long as a condition is `True`.

The **accumulator pattern** starts a total at 0 and adds to it on each pass. You will use it constantly.

````python
# for loop: count from 1 to 5
for day in range(1, 6):
    print("Day", day)

# accumulator pattern: total sales for 4 orders of $10.00 each
total_sales = 0
for order in range(4):
    total_sales = total_sales + 10.00
print("Total sales:", total_sales)
````

````python
# while loop: keep restocking until we have at least 10 books
stock = 2
while stock < 10:
    stock = stock + 3
    print("Restocked. Stock is now", stock)
print("Done restocking.")
````
